```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API'))
MODEL_NAME = "gemini-3.5-flash-lite"
```

```python
lost_and_found_faq = """
------ LADY DOAK COLLEGE — CAMPUS LOST AND FOUND FAQ -------

Q: I lost an item on campus. What should I do?
A: Check the place where you last saw it and ask the staff responsible for that area. You can also contact the college helpdesk.

Q: What information should I provide when asking about a lost item?
A: Describe the item, its colour or identifying features, and where and when you last saw it.

Q: I found someone else's item on campus. What should I do?
A: Hand it to a staff member in the area where you found it, or ask the college helpdesk where to submit it.

Q: I lost something in a classroom. Who should I ask?
A: Ask the department or classroom staff and provide the room number and approximate time you lost it.

Q: I lost something in the library. What should I do?
A: Ask the library staff whether the item has been handed in. Describe the item and where you last had it.

Q: I lost something in the computer lab. What should I do?
A: Inform the lab staff and give them the lab room, workstation location, and a description of the item.

Q: I lost an item during a college event. Who should I contact?
A: Ask the event organisers or staff at the venue. You can also contact the college helpdesk for guidance.

Q: I cannot remember where I lost my item. Can I still ask for help?
A: Yes. Share the places you visited and the approximate time you last had the item.

Q: I lost my student ID card. What should I do?
A: Inform the college office that handles student ID cards and ask about the replacement process. You can also ask whether your card has been handed in.

Q: I found a student ID card. What should I do?
A: Give it to a college staff member or ask the helpdesk where to submit it. Do not post the student's personal details online.

Q: I lost my wallet. What should I do?
A: Ask staff in the places you visited and contact the college helpdesk. If your payment cards were inside, contact your bank or card provider promptly.

Q: I found a wallet. What should I do?
A: Hand it to a college staff member or ask the helpdesk where to submit it. Do not remove or share its contents.

Q: I lost my phone. What details should I provide?
A: Provide its make, model, colour, case or other identifying features, and the last place you remember having it.

Q: I found a phone. Should I try to unlock it?
A: No. Hand it to a college staff member or ask the helpdesk where to submit it.

Q: How can I check whether my item has been found?
A: Ask staff in the area where you lost it. You can also contact the college helpdesk with a description of the item.

Q: Will I need to show that an item belongs to me?
A: The college's item-claim procedure is not listed in this FAQ. Ask the staff handling the item what information they need.

Q: Can my friend collect a found item for me?
A: This FAQ does not specify whether another person can collect it. Ask the staff holding the item before sending someone else.

Q: How long does the college keep found items?
A: The storage period is not listed in this FAQ. Contact the college helpdesk for current information.

Q: Is there a fee to claim a found item?
A: Fee information is not listed in this FAQ. Contact the college helpdesk to confirm.

Q: How can I contact Lady Doak College for help?
A: The college lists helpdesk@ldc.edu.in and telephone numbers
+91 452 2530527 and +91 452 2524575. Its address is
Lady Doak College, Tallakulam, Madurai – 625002, Tamil Nadu.
These are general college contact details, not a dedicated
Lost and Found contact.
"""

assistant_instructions=f"""
You are the Campus Lost and Found Assistant. Help students and staff with
questions about reporting, handing in, and claiming lost items.

Use only the FAQ knowledge base below to answer questions. Respond in a
polite, professional, and concise manner. Give practical steps when the
FAQ provides them.

If the answer is not in the FAQ, say:
"I don't have that information in the FAQ. Please contact the college helpdesk at helpdesk@ldc.edu.in."

Do not invent policies, fees, storage periods, opening hours, locations,
or contact details. Do not ask users to share passwords or other sensitive
information.

FAQ KNOWLEDGE BASE:
{lost_and_found_faq}
"""

support_chat=client.chats.create(
    model=MODEL_NAME,
    config=types.GenerateContentConfig(
      system_instruction=assistant_instructions,
      temperature=0.2,
      max_output_tokens=700,
      thinking_config=types.ThinkingConfig(thinking_level='low')
),
    history=[]
)

print("Hello, welcome!")
print("I'm the Campus Lost and Found FAQ assistant. How can I help you today?")
print("Type 'quit' or 'bye' to exit.\n")

while True:
    user_question = input("You: ").strip()

    if user_question.lower() in ["quit", "bye"]:
        print("Assistant: Goodbye!")
        break

    if user_question == "":
        continue

    assistant_response = support_chat.send_message(user_question)
    print(f"Assistant: {assistant_response.text}\n")
```

```python
for message in support_chat.get_history():
    speaker = "You" if message.role == "user" else "Assistant"
    print(f"{speaker}: {message.parts[0].text}\n")
```
